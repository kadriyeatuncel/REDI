```python
def calculate_average(grades_list):
    if len(grades_list) == 0:
        return 0.0

    total = sum(grades_list)
    average = total / len(grades_list)
    return average

def check_passed(name, average):
    if average >= 4.0:
        status = "passed"
    else:
        status = "failed"
    
    print(f"{name} has an average of {average} and has therefore {status}")

# Test data
my_grades = [2.0, 3.5, 1.5]
name = "Max"

result = calculate_average(my_grades)
check_passed(name, result)

empty_list = []
print(calculate_average(empty_list))
```
